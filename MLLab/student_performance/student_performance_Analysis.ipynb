# Visualization

## Histogram
A histogram shows the distribution of values

```python
plt.figure(figsize=(8,5))

plt.hist(df["age"], bins=10) # divide the values into approximately 10 groups

plt.xlabel("Age")
plt.ylabel("Number of students")
plt.title("Distribution of Student Age")
plt.show()
```

### Histogram of final grade

```python
plt.figure(figsize=(8,5))

plt.hist(df["G3"], bins=10)

plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of students")
plt.title("Distribution of final Grades")

plt.show()
```

### Box Plot

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["G3"])

plt.ylabel("Final Grade")
plt.title("Box Plot of final Grades")

plt.show()
```

### Scatter plot

```python
plt.figure(figsize=(8,5))

plt.scatter(df["studytime"], df["G3"])

plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")
plt.title("Study time VS Final Grade")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["absences"], df["G3"])

plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.title("Absences VS Final Grade")

plt.show()
```

### Bar Chart

```python
gender_average = df.groupby("sex")["G3"].mean()
```

```python
gender_average
```

```python
plt.figure(figsize=(8,5))

plt.bar(gender_average.index, gender_average.values)

plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.title("Average Final Grade by gender")

plt.show()
```

### Correlation Heatmap

```python
plt.figure(figsize=(10,8))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
